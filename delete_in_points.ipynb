```python
import arcpy
```

```python
candidatos = r"D:\Documents\CityEngine\Default Workspace\Area_estudio\models\estudio.gdb\candidatos_cam"
arcpy.management.MakeFeatureLayer(candidatos, "candidatos")
```

```python
arcpy.CreateFeatureclass_management(r"D:\Documents\CityEngine\Default Workspace\Area_estudio\models\estudio.gdb", "candidatos_cam_elevated", "POINT", spatial_reference=candidatos, has_z="ENABLED")
```

```python
with arcpy.da.SearchCursor("candidatos", ["OID@", "SHAPE@XY", "SHAPE@Z"]) as cursor:
    for oid, (x, y), z in cursor:
        with arcpy.da.InsertCursor(r"D:\Documents\CityEngine\Default Workspace\Area_estudio\models\estudio.gdb\candidatos_cam_elevated", ["SHAPE@"]) as c:
            c.insertRow([arcpy.Point(x, y, z + 1000)])
```
